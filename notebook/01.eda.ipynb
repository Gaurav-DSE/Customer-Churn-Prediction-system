## Contract vs Churn

```python
pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100
```

```python
sns.countplot(
    x="Contract",
    hue="Churn",
    data=df
)
plt.show()
```

>Month-to-month customers have a  higher churn rate (42.64%) compared with one-year (11.28%) and two-year customers (2.85%).
>customers with longer-term contracts are considerably more likely to remain with the company

## Numerical feature vs churn

```python
num_cols = ["tenure", "MonthlyCharges", "TotalCharges"]

for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x="Churn", y=col, data=df)
    plt.title(f"{col} vs Churn")
    plt.show()

    print("\n", "="*50)
    print(col)
    print("="*50)
    print(df.groupby("Churn")[col].describe())
```

>customers who churned had substantially lower tenure (median 10 months) compared with customers who stayed (median 38 months), indicating that newer customers are more likely to churn.
>
>Customers who churned had higher monthly charges than customers who stayed, suggesting that higher monthly costs are associated with increased churn.
>
>Churned customers have substantially lower total charges than retained customers. However, this is likely influenced by their shorter tenure, since total charges accumulate over the customer's lifetime.
